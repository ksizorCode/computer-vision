#  Teachable Machine

Teachable Machine es una herramienta web desarrollada por Google, muy facil de usar, que permite a cualquier persona crear modelos de machine learning de forma rápida sin necesidad de programar.

Fue lanzada en 2017 y actualizada en 2019.

https://teachablemachine.withgoogle.com/

## Características Principales
### Accesibilidad Total

- No requiere instalación de software
- Interfaz completamente visual e intuitiva
- No necesita conocimientos de programación
- Funciona directamente desde el navegador web

## Tres Tipos de Modelos

- Imágenes: Reconocimiento y clasificación de objetos, gestos, poses
- Sonidos: Clasificación de audio, música, voces, ruidos
- Poses: Detección y clasificación de posiciones corporales

# Computer Vision con Teachable Machine

La funcionalidad de computer vision de Teachable Machine permite crear clasificadores de imágenes mediante un proceso de tres pasos:
1. Recopilar y Agrupar
    - Crear diferentes clases o categorías
    - Subir imágenes o usar la webcam para capturar ejemplos
    - Se recomienda al menos 10 imágenes por clase
    - Incluir variaciones: diferentes ángulos, iluminación, fondos

2. Entrenar

    - El algoritmo analiza las características de las imágenes
    - Aprende a distinguir entre las diferentes clases
    - Proceso automatizado que toma solo unos minutos
    - Utiliza redes neuronales convolucionales (CNN) simplificadas

3. Probar y Exportar

    - Evaluar el modelo con nuevas imágenes
    - Ajustar agregando más ejemplos si es necesario
    - Exportar para usar en aplicaciones web, móviles o proyectos de programación

```python
import tensorflow as tf
import keras
import google.protobuf

print("TensorFlow:", tf.__version__)
print("Keras:", keras.__version__)
print("Protobuf:", google.protobuf.__version__)
```

```python
import tensorflow.keras
import numpy as np
import cv2

# Disable scientific notation for clarity
np.set_printoptions(suppress=True)

# Load the Keras model
model = tensorflow.keras.models.load_model('converted_keras/keras_model.h5', compile=False)

# Load the labels
class_names = open('converted_keras/labels.txt', 'r').readlines()

# CAMERA setup
camera = cv2.VideoCapture(0) # 0 for default camera, change if you have multiple cameras

# Check if camera opened successfully
if not camera.isOpened():
    print("Error: Could not open camera.")
    exit()

while True:
    # Read a frame from the camera
    ret, image = camera.read()
    if not ret:
        print("Error: Could not read frame from camera.")
        break

    # Resize the raw image to (224, 224) pixels
    image_resized = cv2.resize(image, (224, 224), interpolation=cv2.INTER_AREA)

    # Make the image a numpy array and reshape it to be the appropriate size for the model
    # That is, (1, 224, 224, 3)
    image_array = np.asarray(image_resized, dtype=np.float32).reshape(1, 224, 224, 3)

    # Normalize the image
    normalized_image_array = (image_array / 127.5) - 1

    # Predict the model
    prediction = model.predict(normalized_image_array)
    index = np.argmax(prediction)
    class_name = class_names[index]
    confidence_score = prediction[0][index]

    # Display the prediction on the original frame
    text = f"Class: {class_name[2:].strip()} Confidence: {confidence_score:.2f}"
    cv2.putText(image, text, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

    # Show the image in a window
    cv2.imshow('Webcam Image', image)

    # Listen to the keyboard for the ESC key to exit
    if cv2.waitKey(1) & 0xFF == 27: # 27 is the ASCII for the ESC key
        break

# Release the camera and destroy all windows
camera.release()
cv2.destroyAllWindows()
```
