# Apuntes Computer Vision - Detección de Códigos de Barras

## 1. Introducción al Proyecto

Este código implementa un **sistema de detección y lectura de códigos de barras en tiempo real** utilizando la webcam. Combina técnicas de Computer Vision con librerías especializadas para el reconocimiento de códigos.

## 2. Librerías Utilizadas

### 2.1 OpenCV (cv2)
- **Función**: Procesamiento de imágenes y video en tiempo real
- **Uso en el proyecto**:
  - Captura de video desde webcam
  - Dibujo de elementos gráficos (rectángulos, texto)
  - Visualización de frames
  - Manejo de ventanas

### 2.2 PyZbar
- **Función**: Decodificación de códigos de barras y QR
- **Características**:
  - Detecta múltiples tipos de códigos (QR, Code128, EAN, etc.)
  - Extrae coordenadas del bounding box
  - Decodifica el contenido del código

## 3. Análisis del Código

### 3.1 Inicialización de la Cámara
```python
cap = cv2.VideoCapture(0)
```
- **VideoCapture(0)**: Accede a la cámara principal del sistema
- **Parámetro 0**: Índice de la cámara (0 = cámara por defecto)

### 3.2 Bucle Principal de Captura
```python
while True:
    ret, frame = cap.read()
    if not ret:
        break
```
- **cap.read()**: Captura un frame de la cámara
- **ret**: Booleano que indica si la captura fue exitosa
- **frame**: Array numpy con la imagen capturada

### 3.3 Detección de Códigos
```python
barcodes = pyzbar.decode(frame)
```
- **pyzbar.decode()**: Función principal que analiza la imagen
- **Algoritmo interno**:
  1. Convierte automáticamente a escala de grises
  2. Aplica filtros de mejora de contraste
  3. Busca patrones característicos de códigos
  4. Extrae y decodifica la información

### 3.4 Procesamiento de Resultados
```python
for barcode in barcodes:
    (x, y, w, h) = barcode.rect
    barcode_data = barcode.data.decode("utf-8")
    barcode_type = barcode.type
```
- **barcode.rect**: Coordenadas del bounding box (x, y, ancho, alto)
- **barcode.data**: Datos raw del código (en bytes)
- **barcode.type**: Tipo de código detectado (QR_CODE, CODE_128, etc.)

## 4. Elementos de Visualización

### 4.1 Dibujo del Bounding Box
```python
cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
```
- **Parámetros**:
  - `frame`: Imagen donde dibujar
  - `(x, y)`: Punto superior izquierdo
  - `(x + w, y + h)`: Punto inferior derecho
  - `(0, 255, 0)`: Color BGR (verde)
  - `2`: Grosor del borde

### 4.2 Texto Informativo
```python
cv2.putText(frame, text, (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
```
- **Posición**: `(x, y - 10)` para colocar texto encima del rectángulo
- **Fuente**: `FONT_HERSHEY_SIMPLEX` (fuente estándar)
- **Escala**: `0.6` (tamaño del texto)

## 5. Conceptos de Computer Vision Aplicados

### 5.1 Procesamiento en Tiempo Real
- **Frame Rate**: Procesamiento continuo de frames

### 5.2 Detección de Patrones

````python
import cv2
from pyzbar import pyzbar

# Abrir webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Detectar códigos de barras
    barcodes = pyzbar.decode(frame)

    for barcode in barcodes:
        # Extraer coordenadas del bounding box
        (x, y, w, h) = barcode.rect
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

        # Decodificar datos del código
        barcode_data = barcode.data.decode("utf-8")
        barcode_type = barcode.type

        # Mostrar el texto encima del rectángulo
        text = f"{barcode_data} ({barcode_type})"
        cv2.putText(frame, text, (x, y - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    # Mostrar imagen en ventana
    cv2.imshow("Lector de Códigos de Barras", frame)

    # Pulsar 'q' para salir
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
````
