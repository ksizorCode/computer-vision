```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import urllib.request
import ipywidgets as widgets
from ipywidgets import interact

# URL de la imagen
url = "https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png"

# Agregar cabecera de navegador para evitar error 403
req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
resp = urllib.request.urlopen(req)

# Convertir a array de numpy
image_array = np.asarray(bytearray(resp.read()), dtype=np.uint8)

# Decodificar con OpenCV
image = cv2.imdecode(image_array, cv2.IMREAD_COLOR)

# Convertir a RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Escala de grises
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Función interactiva para aplicar Canny con distintos parámetros
def aplicar_canny(threshold1=100, threshold2=200):
    edges = cv2.Canny(gray, threshold1, threshold2)

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    axes[0].imshow(image_rgb)
    axes[0].set_title("Imagen original")
    axes[0].axis("off")

    axes[1].imshow(gray, cmap="gray")
    axes[1].set_title("Escala de grises")
    axes[1].axis("off")

    axes[2].imshow(edges, cmap="gray")
    axes[2].set_title(f"Bordes Canny ({threshold1}, {threshold2})")
    axes[2].axis("off")

    plt.tight_layout()
    plt.show()

# Widgets interactivos
interact(aplicar_canny, threshold1=(0, 255, 10), threshold2=(0, 255, 10))
```
