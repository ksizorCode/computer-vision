```python
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt
import requests

# URL de la imagen de Lenna
url = "https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png"

# Descargar la imagen y guardarla temporalmente
response = requests.get(url)
with open("lenna.png", "wb") as f:
    f.write(response.content)

# Abrir la imagen desde el archivo local
image = Image.open("lenna.png")

# Convertir a escala de grises
gray = image.convert("L")

# Aplicar desenfoque
blurred = image.filter(ImageFilter.BLUR)

# Mostrar las imágenes
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(image)
axes[0].set_title("Imagen original")
axes[0].axis("off")

axes[1].imshow(gray, cmap="gray")
axes[1].set_title("Escala de grises")
axes[1].axis("off")

axes[2].imshow(blurred)
axes[2].set_title("Desenfoque (BLUR)")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```
