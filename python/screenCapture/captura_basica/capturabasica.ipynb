```python
import cv2
import numpy as np
from PIL import ImageGrab

def apply_threshold_to_screen():
    """
    Captura la pantalla continuamente y aplica un filtro threshold
    """
    print("Presiona 'q' para salir")
    print("Presiona 't' para cambiar entre threshold normal e inverso")
    print("Usa las teclas '+' y '-' para ajustar el valor del threshold")
    
    threshold_value = 127  # Valor inicial del threshold
    inverse = False  # Tipo de threshold
    
    while True:
        # Capturar la pantalla
        screenshot = ImageGrab.grab()
        
        # Convertir a array de numpy
        frame = np.array(screenshot)
        
        # Convertir de RGB a BGR (formato de OpenCV)
        frame = cv2.cvtColor(frame, cv2.COLOR_RGB2BGR)
        
        # Convertir a escala de grises
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        
        # Aplicar threshold
        if inverse:
            _, thresh = cv2.threshold(gray, threshold_value, 255, cv2.THRESH_BINARY_INV)
        else:
            _, thresh = cv2.threshold(gray, threshold_value, 255, cv2.THRESH_BINARY)
        
        # Mostrar información en la imagen
        text = f"Threshold: {threshold_value} | Tipo: {'Inverso' if inverse else 'Normal'}"
        cv2.putText(thresh, text, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 
                    1, (255, 255, 255), 2)
        
        # Mostrar la imagen con threshold
        cv2.imshow('Screen Threshold Filter', thresh)
        
        # Controles de teclado
        key = cv2.waitKey(1) & 0xFF
        
        if key == ord('q'):
            break
        elif key == ord('t'):
            inverse = not inverse
        elif key == ord('+') or key == ord('='):
            threshold_value = min(255, threshold_value + 5)
        elif key == ord('-'):
            threshold_value = max(0, threshold_value - 5)
    
    cv2.destroyAllWindows()

if __name__ == "__main__":
    apply_threshold_to_screen()
```
