```python
Vecinos Cercanos o Kneibghbors Nearest Neighbors (KNN) es un algoritmo de aprendizaje supervisado que se utiliza tanto para clasificación como para regresión. Sin embargo, es más comúnmente utilizado para problemas de clasificación. La idea principal detrás de KNN es que los objetos similares tienden a estar cerca unos de otros en el espacio de características.

El algoritmo KNN funciona de la siguiente manera:
1. **Almacenamiento de Datos**: KNN es un algoritmo basado en instancias, lo que significa que no tiene una fase de entrenamiento explícita. En su lugar, simplemente almacena todos los datos de entrenamiento.
2. **Cálculo de Distancias**: Cuando se presenta un nuevo punto de datos para clasificar, KNN calcula la distancia entre este punto y todos los puntos en el conjunto de datos de entrenamiento. Las métricas de distancia comunes incluyen la distancia euclidiana, la distancia Manhattan y la distancia de Minkowski.
3. **Selección de Vecinos**: Después de calcular las distancias, KNN selecciona los 'k' puntos de datos más cercanos (vecinos) al nuevo punto. El valor de 'k' es un hiperparámetro que debe ser definido por el usuario.
4. **Votación Mayoritaria**: Para la clasificación, KNN asigna la clase más común entre los 'k' vecinos seleccionados al nuevo punto. En el caso de la regresión, KNN calcula el valor promedio (o mediano) de los valores de los 'k' vecinos y lo asigna al nuevo punto.       


```
